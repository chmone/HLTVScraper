```python



url = 'https://www.hltv.org/stats/matches?startDate=2024-06-09&endDate=2024-09-09&matchType=BigEvents&offset={offset}'

proxies = [
    'http://proxy1.example.com:8080',
    'http://proxy2.example.com:8080',
    # Add more proxies as needed
]

# Headers
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3',
    'Referer': 'https://example.com',
    'Accept-Language': 'en-US,en;q=0.9'
}

datas = []
```

```python
# Use a session to maintain cookies and headers
session = requests.Session()
session.headers.update(headers)

for offset in range(0, 220, 50):
    # Rotate proxies
    proxy = {'http': random.choice(proxies), 'https': random.choice(proxies)}
    
    try:
        result = session.get(url.format(offset=offset), proxies=proxy, verify=False)
        if result.status_code == 403:
            print("Access forbidden. Check your headers or authentication.")
            break
        soup = BeautifulSoup(result.text, 'html.parser')
        items = soup.find_all('div', class_='stats-table matches-table no-sort')
        
        for it in items:
            tbody = it.find('tbody')
            rows = tbody.find_all('tr')
            
            for row in rows:
                cols = row.find_all('td')
                col_data = [col.get_text(strip=True) for col in cols]
                datas.append(col_data)
        
        # Add a delay of 2 seconds between requests
        time.sleep(2)
    except requests.exceptions.RequestException as e:
        print(f"Request failed: {e}")
        continue
```
